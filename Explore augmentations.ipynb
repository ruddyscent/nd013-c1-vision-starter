## Rationale

Brightness (`max_delta=0.15`) and contrast (`min_delta=0.8`, `max_delta=1.2`) simulate moderate illumination changes. These operations preserve small objects that aggressive geometric crops might remove. The improved configuration retains the starter horizontal flip and crop and adds these two transforms. The figure isolates their effects; it does not represent the complete training pipeline.

API: [preprocessor.py](https://github.com/tensorflow/models/blob/master/research/object_detection/core/preprocessor.py), [preprocessor.proto](https://github.com/tensorflow/models/blob/master/research/object_detection/protos/preprocessor.proto). Random seeds make the visual examples reproducible. Moderate perturbations cannot substitute for real nighttime data, and performance must be measured on validation data.
